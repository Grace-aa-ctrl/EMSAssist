# 查看 Embedding 矩阵

加载并查看生成的两个 sparse embedding 矩阵。

```python
import os
from scipy import sparse

EMBEDDING_DIR = "/home/xiangling/EMSAssist/embedding"

icd10_matrix = sparse.load_npz(os.path.join(EMBEDDING_DIR, "icd10_code_bag_embeddings.npz"))
word_matrix = sparse.load_npz(os.path.join(EMBEDDING_DIR, "word_bag_embeddings.npz"))
```

## 矩阵基本信息

```python
print("ICD-10 code bag matrix")
print("shape:", icd10_matrix.shape)
print("non-zero entries:", icd10_matrix.nnz)
print("density:", icd10_matrix.nnz / (icd10_matrix.shape[0] * icd10_matrix.shape[1]))

print("\nWord bag matrix")
print("shape:", word_matrix.shape)
print("non-zero entries:", word_matrix.nnz)
print("density:", word_matrix.nnz / (word_matrix.shape[0] * word_matrix.shape[1]))
```
